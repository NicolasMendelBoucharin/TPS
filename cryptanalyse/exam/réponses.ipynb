```python
s = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789 .,?!"
n = len(s)
c=",?jtajY.tcrL0tvtl?20.Y"
k1=13
k2=(64-50*k1)%n

def weakDecipher1(ciphertext: str, k1: int, k2: int):
    s = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789 .,?!"
    n = len(s)
    res = ""
    k1_inv = pow(k1, -1, n)#Merci au "à toute fins utiles"
    for c in ciphertext:
        pos = s.find(c)
        if pos >= 0:
            pos = (k1_inv * (pos - k2)) % n
            res += s[pos]
        else:
            raise ValueError("Unrecognized character: '%s'" % str(c))
    return res

print(weakDecipher1(c, k1, k2))
```

```python

def weakDecipher2(ciphertext, key):
    if len(key) != 6:
        raise ValueError("Expected a list of length 6 (of integers)")
    s = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789 .,?!"
    n = len(s)
    if (len(ciphertext) % 2 != 0):
        ciphertext += "."
    keymatrix=matrix(Zmod(67), [[key[0], key[1]], [key[3], key[4]]]) #la matrice par quoi on multiplie
    decalage=vector(Zmod(67), [key[2], key[5]])#les clés par lesquelles on
    res = ""
    keymatrix_inv = keymatrix.inverse()
    for i in range(0, len(ciphertext), 2):
        y1 = s.find(ciphertext[i])
        y2 = s.find(ciphertext[i+1])
        y = vector(Zmod(67), [y1, y2])
        x = keymatrix_inv * (y - decalage)
        res += s[int(x[0])] + s[int(x[1])]
    return res
key=[47, 40, 26,17, 50, 14]
key_bis=[13, 17, 42, 0, 5, 22]
message = weakDecipher2(tp_code.q2_challenge, key)
messagebis = weakDecipher2(tp_code.q2_challenge, key_bis)
print(message)
print(messagebis)

```
